Data cleaning

```python
# working copy 
df_clean = df.copy()
```

```python
df_clean
```

```python
df_clean.isnull().sum()
```

```python
# removing duplicates records 

print("duplicates before:",df_clean.duplicated().sum())
```

```python
print("duplicates after:",df_clean.duplicated().sum())
```

```python
# standardize column names 
df_clean.columns = df_clean.columns.str.lower()
```

```python
df_clean.columns
```

```python
# removing leading and triling spaces 

object_cols = df_clean.select_dtypes(include ="object").columns

for col in object_cols:
    df_clean[col] = df_clean[col].str.strip()
```

```python
# verify memory usages 

df.info(memory_usage="deep")
```

```python
df_clean.info(memory_usage="deep")
```

```python
# save to cleaned dataset

df_clean.to_csv(
    r"D:\SecurePay-AI\dataset\cleaned\transactions_clean.csv",
    index=False
)
```
